```python
import pandas as pd
import numpy as np
data= {
    'Product':['Laptop','Mobile','Tablet','Headphones','Smartwatch','Camera'],
    'Price':[55000,20000,30000,2000,8000,45000],
    'Units_sold':[25,60,35,80,50,15]

}
df=pd.DataFrame(data)
print(df)
```

```python
#Data Binning
bins1=[0,10000,30000,60000]
labels1=['Low','Medium','High']
df['Price_Category']=pd.cut(df['Price'],bins=bins1,labels=labels1)
print(df)
```

```python
bins2=[0,30,60,90]
labels2=['Low sales', 'Mediuum sales','High sales']
df['Sales_Category']=pd.cut(df['Units_sold'],bins=bins2,labels=labels2)
print(df)
```

```python
#Data formatting
#Convert data type
print(df.dtypes)
df['Units_sold']=df['Units_sold'].astype(float)
print(df.dtypes)
```

```python
#Convert text to upercase
df['Product']=df['Product'].str.upper()
print(df)
```

```python
#Rounding Values
df['Price']=df['Price'].round(2)
print(df)
```

```python
#Additional Operations
#Sort Data
sorted_df=df.sort_values(by='Price')
print(sorted_df)
```

```python
#Display unique values
print(df['Price_Category'].unique())
```

```python
#Dataset: Food Deivery Order
import pandas as pd
import numpy as np
data={
    'Order_ID':['O1','O2','O3','O4','O5','O6','O7','O8'],
    'Order_Value':[250,450,120,600,300,150,500,220],
    'Delivery_Time':[30,45,15,60,25,10,55,35],
    "Customer_Rating":[4,3,2,5,4,3,2,5],
    'Distance_KM':[10,15,5,20,12,8,18,10]
}
df=pd.DataFrame(data)
print(df)
```

```python
#Bidding Order Values
bin=[0,200,400,700]
labels1=['Low Order','Medium Order','High Order']
df['Order_Value_Category']=pd.cut(df['Order_Value'],bins=bin,labels=labels1)
print(df)
```

```python
#Binning delivery Time
bins2= [0,30,45,70]
labels2=['Fast Delivery','Average Delivery','Slow Delivery']
df['Delivery_Time_Category']=pd.cut(df['Delivery_Time'],bins=bins2,labels=labels2)
print(df)
```

```python
#Data formating examples
#convert distance to float
df['Distance_KM']=df['Distance_KM'].astype(float)
print(df.dtypes)
```

```python
#Convert Order_ID TO UPPERCASE
df['Order_Value_Category']=df['Order_Value_Category'].str.upper()
print(df)
```

```python
#Check datatypes
print(df.dtypes)
```

```python
#Unique categories
print(df['Order_Value_Category'].unique())
df
```
